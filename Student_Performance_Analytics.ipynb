# Student Performance Analytics System

A NumPy-based project to analyze student performance using data stored in a CSV file.

```python
import os

print(os.getcwd())
print(os.listdir())
```

Import Libraries

```python
import csv
import numpy as np
```

Create Dataset

```python
# Student Performance Dataset
data = [
    ["Student_ID", "Maths", "Physics", "English", "Computer_Science"],
    [101, 85, 72, 90, 92],
    [102, 65, 88, 76, 70],
    [103, 92, 95, 89, 97],
    [104, 55, 61, 70, 58],
    [105, 78, 84, 82, 88]
]

# Create CSV file
with open("students.csv", "w", newline="") as file:
    writer = csv.writer(file)
    writer.writerows(data)

print("students.csv created successfully!")
```

Load Dataset

```python
# Load data from CSV file
students_data = np.loadtxt(
    "students.csv",
    delimiter=",",
    skiprows=1,
    dtype="int64"
)

# Subject names
subjects = np.array([
    "Maths",
    "Physics",
    "English",
    "Computer Science"
])

print("Dataset loaded successfully!")
print(students_data)
```

Understand the Dataset

```python
# Get dataset information
total_students = students_data.shape[0]
total_subjects = students_data.shape[1] - 1

print("===== DATASET INFORMATION =====")

print(f"Number of students: {total_students}")
print(f"Number of subjects: {total_subjects}")
print(f"Total number of values: {students_data.size}")
print(f"Data type: {students_data.dtype}")

print(f"Student IDs: {students_data[:, 0]}")
print(f"Marks of Student 103: {students_data[2, 1:]}")
print(f"Maths marks: {students_data[:, 1]}")
```

Subject Analysis

```python
# Select only subject marks
subject_marks = students_data[:, 1:]

# Calculate average of each subject
subject_average = np.mean(subject_marks, axis=0)

# Highest and lowest marks in the entire dataset
highest_marks = np.max(subject_marks)
lowest_marks = np.min(subject_marks)

# Highest and lowest marks for each subject
subject_highest_marks = np.max(subject_marks, axis=0)
subject_lowest_marks = np.min(subject_marks, axis=0)


print("===== SUBJECT ANALYSIS =====")

for i in range(total_subjects):
    print(
        f"{subjects[i]}: "
        f"Average = {subject_average[i]:.2f}, "
        f"Highest = {subject_highest_marks[i]}, "
        f"Lowest = {subject_lowest_marks[i]}"
    )

print(f"\nHighest mark in dataset: {highest_marks}")
print(f"Lowest mark in dataset: {lowest_marks}")
```

Student Analysis

```python
# Calculate average marks of each student
student_average = np.mean(subject_marks, axis=1)

# Find student with highest average
top_student_index = np.argmax(student_average)

top_student_id = students_data[top_student_index, 0]
top_student_average = student_average[top_student_index]


print("===== STUDENT ANALYSIS =====")

print("Average marks of all students:")

for i in range(total_students):
    print(
        f"Student {students_data[i, 0]}: "
        f"{student_average[i]:.2f}"
    )

print(
    f"\nTop Student: {top_student_id} "
    f"with an average of {top_student_average:.2f}"
)
```

Pass / Fail Analysis

```python
# Define pass mark
pass_mark = 70

# Create pass/fail mask
pass_mask = student_average >= pass_mark

# Count passed and failed students
passed_students = np.sum(pass_mask)
failed_students = total_students - passed_students

# Calculate pass percentage
pass_percentage = (passed_students / total_students) * 100

# Get IDs of failed students
fail_student_ids = students_data[~pass_mask, 0]


print("===== PASS / FAIL ANALYSIS =====")

print(f"Pass mark: {pass_mark}")
print(f"Students passed: {passed_students}")
print(f"Students failed: {failed_students}")
print(f"Pass percentage: {pass_percentage:.2f}%")
print(f"Failed student IDs: {fail_student_ids}")
```

Hardest Subject

```python
# Find subject with lowest average
hardest_subject_index = np.argmin(subject_average)

hardest_subject = subjects[hardest_subject_index]
hardest_subject_average = subject_average[hardest_subject_index]


print("===== SUBJECT DIFFICULTY =====")

print(
    f"Hardest subject: {hardest_subject}"
)

print(
    f"Average marks: {hardest_subject_average:.2f}"
)
```

Student Consistency

```python
# Calculate standard deviation for each student
student_std = np.std(subject_marks, axis=1)


print("===== STUDENT CONSISTENCY =====")

for i in range(total_students):
    print(
        f"Student {students_data[i, 0]}: "
        f"Standard Deviation = {student_std[i]:.2f}"
    )
```

Student Ranking

```python
# Sort students based on average marks
ranking_indices = np.argsort(student_average)[::-1]

ranked_student_ids = students_data[ranking_indices, 0]
ranked_averages = student_average[ranking_indices]


print("===== STUDENT RANKING =====")

for i in range(total_students):
    print(
        f"{i + 1}. Student {ranked_student_ids[i]} "
        f"- Average: {ranked_averages[i]:.2f}"
    )
```

Final Summary

```python
print("==========================================")
print("       STUDENT PERFORMANCE SUMMARY")
print("==========================================")

print(f"Total Students       : {total_students}")
print(f"Total Subjects       : {total_subjects}")
print(f"Top Student          : {top_student_id}")
print(f"Top Student Average  : {top_student_average:.2f}")
print(f"Students Passed      : {passed_students}")
print(f"Students Failed      : {failed_students}")
print(f"Pass Percentage      : {pass_percentage:.2f}%")
print(f"Hardest Subject      : {hardest_subject}")
print(f"Hardest Subject Avg  : {hardest_subject_average:.2f}")

print("==========================================")
```
